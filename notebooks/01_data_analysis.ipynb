```python
X = df.drop(columns=["selling_price"])
y = df["selling_price"]

print("Input features:", X.shape)
print("Target:", y.shape)
```

```python
categorical_columns = [
    "vehicle_type",
    "brand",
    "model",
    "fuel_type",
    "transmission",
    "city",
    "body_type"
]

print(categorical_columns)
```

```python
X_encoded = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True
)

print("Original features:", X.shape)
print("Encoded features:", X_encoded.shape)
```

```python
X_encoded.head()
```

```python
print("Original features:", X.shape)
print("Encoded features:", X_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
rf_predictions = rf_model.predict(X_test)

print(rf_predictions[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, rf_predictions)

rmse = np.sqrt(
    mean_squared_error(y_test, rf_predictions)
)

r2 = r2_score(y_test, rf_predictions)

print("Random Forest Results")
print("---------------------")
print("MAE :", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 4))
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=5,
    random_state=42
)

gb_model.fit(X_train, y_train)

print("Gradient Boosting model trained successfully!")
```

```python
gb_predictions = gb_model.predict(X_test)

print(gb_predictions[:10])
```

```python
gb_mae = mean_absolute_error(y_test, gb_predictions)

gb_rmse = np.sqrt(
    mean_squared_error(y_test, gb_predictions)
)

gb_r2 = r2_score(y_test, gb_predictions)

print("Gradient Boosting Results")
print("-------------------------")
print("MAE :", round(gb_mae, 2))
print("RMSE:", round(gb_rmse, 2))
print("R²  :", round(gb_r2, 4))
```

```python
results = pd.DataFrame({
    "Model": ["Random Forest", "Gradient Boosting"],
    "MAE": [mae, gb_mae],
    "RMSE": [rmse, gb_rmse],
    "R2 Score": [r2, gb_r2]
})

results
```
